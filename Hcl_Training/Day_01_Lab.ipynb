```python

units = int(input("Enter units sold: "))
price = float(input("Enter price per unit: ₹"))

target = float(input("Enter sales target: ₹"))

revenue = units * price

if target == 0:
    print("Sales target cannot be zero.")
else:
    achievement = revenue / target

    print(f"Revenue: ₹{revenue:,.2f}")
    print(f"Target achievement: {achievement:.0%}")
```

```python

# Function to get inputs from the user
def get_inputs():
    while True:
        try:
            units = input("Enter units sold (or 'end' to stop): ")
            if units.lower() == "end":
                return None

            price = input("Enter price per unit (or 'end' to stop): ")
            if price.lower() == "end":
                return None

            target = input("Enter sales target (or 'end' to stop): ")
            if target.lower() == "end":
                return None

            units = int(units)
            price = float(price)
            target = float(target)

            return units, price, target

        except ValueError:
            print("Invalid input! Please enter valid numbers.")
            print("Please try again.\n")


# Function to calculate sales performance
def sales_performance(units, price, target):
    revenue = units * price

    try:
        achievement = revenue / target

        print(f"Revenue: ₹{revenue:,.2f}")
        print(f"Target achievement: {achievement:.0%}")

        return True

    except ZeroDivisionError:
        print("Sales target cannot be zero.")
        return False


# Main loop to repeat the program
while True:
    data = get_inputs()

    if data is None:
        print("Program stopped. Goodbye!")
        break

    units, price, target = data

    sales_performance(units, price, target)
```
