```python

units = int(input("Enter units sold: "))
price = float(input("Enter price per unit: ₹"))

target = float(input("Enter sales target: ₹"))

revenue = units * price

if target == 0:
    print("Sales target cannot be zero.")
else:
    achievement = revenue / target

    print(f"Revenue: ₹{revenue:,.2f}")
    print(f"Target achievement: {achievement:.0%}")
```
