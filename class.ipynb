```python
import numpy as np

Data = np.array([
    [101, 50],
    [102, 65],
    [103, 80],
    [104, 45],
    [105, 90]
])

# Separate columns
ids = Data[:, 0]
marks = Data[:, 1]

# (a) Average marks
avg = marks.mean()

# (b) Add 5 marks bonus (keep IDs same)
bonus_data = np.column_stack((ids, marks + 5))

# (c) Filter students scoring < 60 (full rows)
filtered_students = Data[marks < 60]

print("Average Marks:", avg)
print("Data after bonus:\n", bonus_data)
print("Students scoring < 60:\n", filtered_students)
```
