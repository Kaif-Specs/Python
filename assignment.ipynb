```python
import matplotlib.pyplot as plt

studyhours = [1,2,3,4]
marks = [10,20,25,30]

plt.scatter(studyhours,marks)
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("Study Relationship")
plt.show()
```

```python
import matplotlib.pyplot as plt

month = ["jan","feb","march","april"]
sales = [100,200,250,300]

plt.bar(month,sales)
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("Monthly Sales")
plt.show()
```

```python
import matplotlib.pyplot as plt

Student = [20,40,30,60]
Courses = ['MCA','B.Tech','MBA','BCOM']

plt.plot(Student,Courses)
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("Courses opted by Students")
plt.show()
```
